```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/NLP/train.txt/train.txt',sep=';',header=None,names=['text','emotion'])
```

```python
unique_emotion = data['emotion'].unique()
emotional_number ={}
i = 0
for emo in unique_emotion:
  emotional_number[emo] = i
  i += 1
```

```python
data['emotion'] = data['emotion'].map(emotional_number)
```

# Convert into lowercase

```python
data['text'] = data['text'].apply(lambda x:x.lower())
```

```python
data.head(

)
```

# Remove Punctuation

```python
import string
def remove_punctuation(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
data['text'] = data['text'].apply(remove_punctuation)
```

# Remove Number

```python
def remove_number(txt):
  new = ""
  for i in txt:
    if not i.isdigit():
      new = new+i
  return new
data['text'] = data['text'].apply(remove_number)
```

```python
data.head()
```

# Removing Emojies

```python
def remove_emojies(txt):
  new = ""
  for i in txt:
    if i.isascii():
      new += i
  return new
data['text'] = data['text'].apply(remove_emojies)
```

# Remove Stopwords

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = stopwords.words('english')
len(stop_words)
```

```python
def remove(txt):
  words = word_tokenize(txt)
  clean_txt = []
  for i in words:
    if not i in stop_words:
      clean_txt.append(i)
  return " ".join(clean_txt)
```

```python
nltk.download('punkt_tab')
```

```python
data['text'] = data['text'].apply(remove)
```
